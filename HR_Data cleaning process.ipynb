```python
#Drop the useless Columns
df = df.drop(columns=["EmployeeCount", "Over18", "StandardHours"])
```

Feature Engineering

```python
education_map = {
    1: "Below College",
    2: "College",
    3: "Bachelor",
    4: "Master",
    5: "Doctor"
}
df["EducationLevel"] = df["Education"].map(education_map)
```

```python
environment_map = {
    1: "Low",
    2: "Medium",
    3: "High",
    4: "Very High"
}
df["EnvironmentSatisfactionLevel"] = (
    df["EnvironmentSatisfaction"].map(environment_map)
)
```

```python
job_involvement_map = {
    1: "Low",
    2: "Medium",
    3: "High",
    4: "Very High"
}
df["JobInvolvementLevel"] = (
    df["JobInvolvement"].map(job_involvement_map)
)
```

```python
joblevel_map = {
    1: "Junior",
    2: "Mid-Level",
    3: "Senior",
    4: "Senior",
    5: "Senior"
}

df["JobLevelCategory"] = df["JobLevel"].map(joblevel_map)
```

```python
def companies_category(x):
    if x == 0:
        return "No Previous Companies"
    elif x <= 2:
        return "1-2 Companies"
    elif x <= 4:
        return "3-4 Companies"
    else:
        return "5+ Companies"

df["PreviousCompaniesCategory"] = df["NumCompaniesWorked"].apply(
    companies_category
)
```

```python
def experience_category(x):
    if x <= 2:
        return "0-2 Years"
    elif x <= 5:
        return "3-5 Years"
    elif x <= 10:
        return "6-10 Years"
    elif x <= 20:
        return "11-20 Years"
    else:
        return "21+ Years"

df["TotalExperienceCategory"] = df["TotalWorkingYears"].apply(
    experience_category
)
```

```python
worklife_map = {
    1: "Bad",
    2: "Good",
    3: "Very Good",
    4: "Excellent"
}
df["WorkLifeBalanceLevel"] = df["WorkLifeBalance"].map(worklife_map)
```

```python
job_satisfaction_map = {
    1: "Low",
    2: "Medium",
    3: "High",
    4: "Very High"
}
df["JobSatisfactionLevel"] = df["JobSatisfaction"].map(job_satisfaction_map)
```

```python
relationship_satisfaction_map = {
    1: "Low",
    2: "Medium",
    3: "High",
    4: "Very High"
}
df["RelationshipSatisfactionLevel"] = df["RelationshipSatisfaction"].map(relationship_satisfaction_map)
```

```python
environment_satisfaction_map = {
    1: "Low",
    2: "Medium",
    3: "High",
    4: "Very High"
}
df["EnvironmentSatisfactionLevel"] = df["EnvironmentSatisfaction"].map(environment_satisfaction_map)
```

```python
job_involvement_map = {
    1: "Low",
    2: "Medium",
    3: "High",
    4: "Very High"
}
df["JobInvolvementLevel"] = df["JobInvolvement"].map(job_involvement_map)
```

```python
def age_category(age):
    if age <= 25:
        return "18-25"
    elif age <= 35:
        return "26-35"
    elif age <= 45:
        return "36-45"
    elif age <= 55:
        return "46-55"
    else:
        return "56+"
df["AgeCategory"] = df["Age"].apply(age_category)
```
